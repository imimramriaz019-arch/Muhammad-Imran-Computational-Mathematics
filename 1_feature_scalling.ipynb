# The feature scaling is only used on the `NUMERIC` values:

## 1. Min Max Scalling

```python
import pandas as pd
from sklearn.preprocessing import MinMaxScaler
# sample data
data = {'awen_numbers': [10, 25, 38, 49, 56, 62, 70, 80, 90, 100]}
df = pd.DataFrame(data)
df
```

## 2. Scale the data uisng MinMax Scalar

```python
scaler = MinMaxScaler()
# Fit the scaler to the data
df['awen_numbers_scaled'] = scaler.fit_transform(df[['awen_numbers']])
df.head()
```

## 2. Standard Scalar or Z-Score normalization

```python
import pandas as pd
from sklearn.preprocessing import StandardScaler
# sample data
data = {'awen_numbers': [10, 25, 38, 49, 56, 62, 70, 80, 90, 100]}
df = pd.DataFrame(data)
df
```

```python
scaler = StandardScaler()
# Fit the scaler to the data
df['awen_numbers_scaled'] = scaler.fit_transform(df[['awen_numbers']])
df
```

## 3. Robust Scaler

```python
import pandas as pd
from sklearn.preprocessing import RobustScaler
# sample data
data = {'awen_numbers': [10, 25, 38, 500,70,88]}
df = pd.DataFrame(data)
```

```python
scaler = RobustScaler()
# Fit the scaler to the data
df['awen_numbers_scaled'] = scaler.fit_transform(df[['awen_numbers']])
df
```

## Logrithmic scaling/normalization

```python
import pandas as pd
import numpy as np
# random data  with outliers
data = {'awen_numbers': [1000, 20000, 3000, 500000,7000,8000]}
df = pd.DataFrame(data)
# Log transformation
df['awen_numbers_log'] = np.log(df['awen_numbers'])
df['awen_numbers_log10'] = np.log10(df['awen_numbers'])
df['awen_numbers_log2'] = np.log2(df['awen_numbers'])
df.head()
```
